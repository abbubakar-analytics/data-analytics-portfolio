## Data Preparing for Machine Learning

#### Separating Target and Features

The next step is to prepare the dataset by separating the target variable from the input features. In our Sales Effectiveness project,Status represents the outcome of lead, so it will be our target variable.Before making any transformations, we will create separate variables for the target and features while keeping the original df unchanged.

```python
# Separating the target variable from the feature variables
X = df.drop('Status',axis=1)
Y = df['Status']

print("Features and target variable separated successfully!")
```

#### Removing Non-Predictive Identifier and Contact Columns

Some columns in the dataset are identifiers or contact details rather than meaning predictors of sales lead outcomes.Product_ID,Mobile,and Email are not suitable as direct machine-learning features because they do not provide meaningful predictive information. Removing these columns helps the model focus on relevant business-related information while reducing unnecessary complexity.

```python
# Removing identifier and contact information from the features

X = X.drop(['Product_ID', 'Mobile', 'EMAIL'], axis = 1)

print("Non-predictive identifier and contact columns removed successfully!")
```

### Converting the Date Column into Useful Features

The Created column is already converted to datetime format, but machine-learning models generally work better when dates are represented as meaningful numerical features.Therefore, we will extract the month and day of the week from Created. These features can help the model identify whether lead outcomes vary according to when leads were created. After extracting these useful features, we will remove the orginal Created column from X.

```python
# Extracting useful features from the Created date

X['Created_Month'] = X['Created'].dt.month
X['Created_DayofWeek'] = X['Created'].dt.dayofweek

# Removing the orginal date column
X = X.drop('Created', axis = 1)

print("Date features extracted and orginal Created column removed successfully!")
```

### Encoding Categorical Features

The remaining input features such as Source, Sales_Agent, Location and Delivery_Mode contain categorical values represented as text.
Machine-learning algorithms generally require numerical input, so these categorical features need to be converted into numerical form. We will use One-Hot Encoding, which creates separate binary columns for each category.

```python
# Encoding categorical features using One-Hot Encoding

X = pd.get_dummies(X, columns = ['Source', 'Sales_Agent', 'Location', 'Delivery_Mode'], drop_first =True)

print('Categorical features encoded successfully!')
```

## Target Variable Preparation

The original Status column contains several detailed lead outcomes. To match the project objective, these outcomes will be consolidated into two meaningful categories: High Potential and Low Potential. This converts the problem into a binary classification task for the machine-learning model

```python
# Creating Lead Category from Status

status_to_category = {'JUNK LEAD' : 'Low Potential', 'NOT RESPONDING' : 'Low Potential',
                      'JUST ENQUIRY' : 'Low Potential', 'IN PROGRESS NEGATIVE' : 'Low Potential',  'LOST' : 'Low Potential',
                      'CONVERTED' : 'High Potential', 'POTENTIAL' : 'High Potential', 'LONG TERM' : 'High Potential',
                      'IN PROGRESS POSITIVE' : 'High Potential', 'OPEN' : 'High Potential'}

df['Lead_Category'] = (df['Status'].str.upper().str.strip().map(status_to_category))
print('Lead Category created successfully!')
```

### Verifying Lead_Category

Before using the target variable for machine learning, we need to verify that every record has been correctly assigned to either High Potential or Low Potential. This quality check ensures that no records were left uncategorized during the mapping process.

```python
# Verifying Lead Category values

print(df['Lead_Category'].value_counts())

print("\nMissing values in Lead_Category :")
print(df['Lead_Category'].isnull().sum())
```

## Train-Test Split

The prepared dataset will now be divided into two separate sets: a training set and a testing set. The training set will be used to teach the machine-learning model the patterns in the sales data, while the testing set will be used to evaluate the model on unseen data. We will use an 80:20 split, with 80% of the records for training and 20% of the records for testing. 

The target variable represents the project objective of predicting whether a sales lead is High Potential or Low Potential.

```python
# Preparing the final target variable

Y = df['Lead_Category']

# Splitting the dataset into training and testing sets
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.20, random_state = 42, stratify = Y)

print("Data successfully split into training and testing sets!")
print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", Y_train.shape)
print("Testing target shape:", Y_test.shape)
```

### Target Encoding

The target variable Lead_Category contains two categorical classes: High Potential and Low Potential.Since machine-learning models require numerical target values,Label Encoding is used to convert these categories into numerical labels.

```python
# Encoding the binary target variable

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

Y_train_encoded = label_encoder.fit_transform(Y_train)
Y_test_encoded = label_encoder.transform(Y_test)

print("Target variable encoded successfully!")
```

```python
# Checking the encoded target classes

print(label_encoder.classes_)
```

## Machine Learning Model Building

#### Decision Tree Model

A Decision Tree Classifier is used to predict the lead category based on the available lead-related features. The model learns decision rules from the training data and classifies each lead as either High Potential or Low Potential.

```python
# Building the Decision Tree model

from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(random_state = 42)
decision_tree.fit(X_train, Y_train_encoded)

print("Decision Tree model trained successfully!")
```

### Decision Tree Prediction

The trained Decision Tree model is used to predict the Lead Category for the testing dataset. These prediction will later be compared with the actual target values to evaluate the model's performance.

```python
# Making prediction on the test data

Y_pred = decision_tree.predict(X_test)

print("Prediction generated successfully!")
```

#### Model Evaluation - Accuracy

Accuracy measures the percentage of test records that were classified correctly by the model. It provides an initial understanding of the Decision Tree's performance on unseen data.

```python
# Checking the model accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test_encoded, Y_pred)

print("Decision Tree Accuracy:", accuracy)
```

#### Decision Tree -Confusion Matrix

A confusion matrix shows how many High Potential and Low Potential leads were classified correctly or incorrectly by the Decision Tree model. It helps us understand the types of predictions made by the model.

```python
# Creating the confusion matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test_encoded, Y_pred)
print(cm)
```

#### Classification Report

The classification report is used to evaluate the performance of the Decision Tree model using precision, recall, and F1-score for each lead category.

```python
# Generating the classification report

from sklearn.metrics import classification_report

print(classification_report(Y_test_encoded, Y_pred))
```

### Random Forest Model

A Random Forest Classifier is used to predict whether a sales lead is High Potential or Low Potential. It combines multiple decision trees to improve prediction performance and reduce the limitations of using a single decision tree.

```python
# Building the Random Forest model

from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(random_state = 42)

random_forest.fit(X_train, Y_train_encoded)
print("Random Forest model trained successfully!")
```

```python
# Making predictions with Random Forest

Y_pred_rf = random_forest.predict(X_test)

print("Random Forest predictions generated successfully!")
```

#### Random Forest Accuracy

```python
# Checking Random Forest accuracy

accuracy_rf = accuracy_score(Y_test_encoded, Y_pred_rf)
print("Random Forest Accuracy:", accuracy_rf)
```
