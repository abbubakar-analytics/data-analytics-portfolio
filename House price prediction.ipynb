## Data Cleaning

Identifying and removing duplicate records and correcting the any inconsistencies in dataset

```python
### Removing duplicates records from the dataset
print('Before:', df.shape)

df.drop_duplicates(inplace=True)

print('After:', df.shape)
```

#### Since there is no duplicate values in dataset. 

## Handling Missing Values

Detecting the missing values in the dataset

```python
## Checking missing values in the dataset
missing = df.isnull().sum()

missing = missing[missing>0]

print(missing.sort_values(ascending=False))
```

```python
## Filling missing values

## Numerical columns -> Median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

## Categorical columns -> Mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Done")
```

### Exploratory Data Analysis(EDA)

Performing visual and statistical analysis to discover patterns, trends, and relationship in the data

```python
## Visualizing the distribution of house prices

plt.figure(figsize=(8,5))
sns.histplot(df['SalePrice'], kde=True)
plt.title('Sale Price Distribution')
plt.show()
```

#### Observation

### Correlation Heatmap

```python
## Relationships between numerical features in dataset

plt.figure(figsize=(12,8))

sns.heatmap(df.select_dtypes(include=np.number).corr(), cmap='coolwarm')

plt.show()
```

#### Observation

- The heatmap represents the correlation among numerical features in the dataset.
- Strong positive correlations are shown by dark red colors, while negative correlation are shown by blue colors.
- Features such as OverallQual, TotalBsmtSF, and YearBuilt show relatively strong positive relationships with SalePrice.
- Correlation analysis helps identify important variables for feature engineering and feature selection.
