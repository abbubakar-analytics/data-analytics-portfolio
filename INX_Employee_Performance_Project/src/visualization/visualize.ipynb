# INX Future Inc.Employee Performance Analysis

## Data Visualization

### Objective

To visualize employee performance data and present the key findings from the analysis, including diffrences in performance across departments and the factors associated with employee performance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "../../data/raw/INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls"
df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:",df.shape)
```

### Department-wise Employee Performance

This chart compares the average employee performance rating across departments.

```python
department_performance = df.groupby('EmpDepartment')['PerformanceRating'].mean().sort_values(ascending = False)

department_performance.plot(kind = 'bar',figsize = (10,5))
plt.title('Average Performance Rating')
plt.xlabel('Department')
plt.ylabel('Average Performance Rating')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

### Observation

The bar chart compares average employee performance ratings across depatments. Development has the highest average rating, while Finance has the lowest, consistent with the department-wise analysis.

### Top Factors associated Employee Performance

This chart compares the mutual information scores of the three highest-ranked factors associated with employee performance.

```python
top_factors = pd.Series({'EmpLastSalaryHikePercent' : 0.151473, 'EmpEnvironmentsatisfaction' : 0.137456, 'YearsSinceLastPromotion' : 0.070989})

top_factors.plot(kind = 'bar',figsize= (8,5))
plt.title('Top Three Factors Associated with Employee Performance')
plt.xlabel('Employee Factors')
plt.ylabel('Mutual Information score')
plt.xticks(rotation = 30, ha = 'right')
plt.tight_layout()
plt.show()
```

### Observation

Employee Last Salary Hike Percent has the highest mutual information score, followed by Employee Environment Satisfaction and years Since Last Promotion. These are the three highest-ranked factors in the analysis.
