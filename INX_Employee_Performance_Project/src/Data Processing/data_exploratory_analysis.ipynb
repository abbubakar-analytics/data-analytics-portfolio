# INX Future Inc.-Employee Performance Analysis

## 2. Data Exploratory Analysis

### Loading the Employee Dataset

The employee dataset is loaded for further exploratory analysis.

```python
import pandas as pd

file_path = "../../data/raw/INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls"
df = pd.read_excel(file_path)

print("Dataset shape:", df.shape)
```

```python
df.head()
```

### Department-wise Performance Analysis

This analysis compares employee performance ratings across different departments.

```python
df.groupby('EmpDepartment')['PerformanceRating'].mean().sort_values(ascending = False)
```

### Observation

Development and Data Science have the highest average performance ratings, while Finance has the lowest. This shows differences in average performance across departments.

### Factors Affecting Employee Performance

This analysis examines the relationship between employee-related factors and performance ratings to identify factors that may influence employee performance.

```python
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import OrdinalEncoder

X = df.drop(columns = ['PerformanceRating', 'EmpNumber'])
y = df['PerformanceRating']

categorical_cols = X.select_dtypes(include = 'object').columns
X_encoded = X.copy()
X_encoded[categorical_cols] = OrdinalEncoder().fit_transform(X[categorical_cols])

discrete_mask = [col in categorical_cols for col in X_encoded.columns]

scores = mutual_info_classif(X_encoded, y, discrete_features = discrete_mask, random_state = 42)

pd.Series(scores,index = X_encoded.columns).sort_values(ascending = False).head(10)
```

### Observation

Employee Last Salary Hike Percent has the highest mutual information score, followed by Employee Environment Satisfaction and Years Since Last Promotion. These are the three highest-ranked factors in this analysis and should be examined further to understand their relationship with employee performance.

### Correlation Analysis of Important Factors

This analysis examines the correlation between the selected numerical factors and employee performance ratings.

```python
df[['EmpLastSalaryHikePercent', 'EmpEnvironmentSatisfaction', 'YearsSinceLastPromotion', 'PerformanceRating']].corr()
```

### Observation

Employee Environment Satisfaction has the strongest positive correlation with performanceRating, followed by Employee Last Salary Hike Percent. Years Since Last Promotion shows a weak negative correlation with PerformanceRating. These results describe relationships in the dataset and do not establish cause and effect.
