# INX Future Inc. -Employee Performance Analysis

## Employee Performance Prediction

## Objective

To use the trained machine learning model to predict employee performnace ratings based on employee-related factors.

## Load the Trained Model and Encoder

```python
import joblib

model = joblib.load("employee_performance_model.pkl")
encoder = joblib.load("employee_performance_encoder.pkl")

print("Trained model and encoder loaded successfully!")
```

```python
import pandas as pd

file_path = "../../data/raw/INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls"
df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
```

## Prepare Employee Data for Prediction

```python
X = df.drop(columns = ['PerformanceRating', 'EmpNumber'])

categorical_cols = X.select_dtypes(include = 'object').columns
X[categorical_cols] = encoder.transform(X[categorical_cols])

print("Employee features prepared successfully!")
```

## Predict Employee Performance

```python
y_pred = model.predict(X)

df['PredictedPerformanceRating'] = y_pred

print("Employee performance predictions generated successfully!")
print(df[['PredictedPerformanceRating']].head())
```
