# INX Future Inc. -Employee Performance Analysis

## Model Training and Evaluation

### Objective

To train and evaluate a machine learning model that predicts employee performance ratings using the available employee data.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
file_path = "../../data/raw/INX_Future_Inc_Employee_Performance_CDS_Project2_Data_V1.8.xls"
df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Dataset shape :", df.shape)
```

### Preparing Features and Target

The employee data is divided into input features and the target variable. PerformanceRating is the target we want the model to predict, and EmpNumber is excluded because it is an employee identifier.

```python
X = df.drop(columns = ['PerformanceRating', 'EmpNumber'])
y = df['PerformanceRating']

print("Features shape :", X.shape)
print("Target shape :", y.shape)
```

### Splitting the Dataset

The dataset is divided into training and testing sets. The training data will be used to train the model, while the testing data will be used to evaluate its performance.

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.20, random_state = 42)

print("Training data shape :", X_train.shape)
print("Testing data shape :", X_test.shape)
```

### Encoding Categorical Features

Categorical employee information is converted into numerical values so it can be used by the machine learning model.

```python
categorical_cols = X_train.select_dtypes(include = 'object').columns

encoder = OrdinalEncoder(handle_unknown = 'use_encoded_value', unknown_value = -1)

X_train[categorical_cols] = encoder.fit_transform(X_train[categorical_cols])
X_test[categorical_cols] = encoder.transform(X_test[categorical_cols])

print("Categorical features encoded successfully!")
```

### Training the Machine Learning Model

A Random Forest Classifier is trained using the encoded employee training data to predict employee performance ratings.

```python
model = RandomForestClassifier(random_state = 42)
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

### Model Evaluation

The trained model is used to predict performance ratings for the testing data. The predictions are compared with the actual ratings to evaluate the model.

```python
y_pred = model.predict(X_test)

print("Model Accuracy :", accuracy_score(y_test, y_pred))
print("\nClassification Report :\n")
print(classification_report(y_test, y_pred))
```

```python
print("Confusion Matrix :")
print(confusion_matrix(y_test, y_pred))
```

### Feature Importance

Feature importance is used to identify which employee factors contribute most to the Random Forest model's predictions.

```python
feature_importance = pd.Series(model.feature_importances_, index = X_train.columns).sort_values(ascending = False)

print("Top 10 Important Features:")
print(feature_importance.head(10))
```

## Observation

Employee Environment Satisfaction has the highest feature importance, followed by Employee Last Salary Hike Percent and Years Since Last Promotion. These are the three most important features identified by the Random Forest model for predicting employee performance ratings.

### Save the Trained Model

```python
import joblib

joblib.dump(model, "employee_performance_model.pkl")

print("Trained model saved successfully!")
```

### Save the Categorical Encoder

```python
joblib.dump(encoder, "employee_performance_encoder.pkl")

print("Categorical encoder saved successfully!")
```
